```python
import os
from typing import TypedDict, List, Dict, Any
from google.colab import userdata
from openai import OpenAI

# Загружаем ключ OpenRouter из секретов Colab
OPENROUTER_API_KEY = userdata.get('OPENROUTER_API_KEY')
OPENROUTER_BASE_URL = "https://openrouter.ai/api/v1"

client = OpenAI(
    base_url=OPENROUTER_BASE_URL,
    api_key=OPENROUTER_API_KEY,
)

# Модель для генерации (выберите любую доступную, например бесплатную или дешёвую)
LLM_MODEL = "deepseek/deepseek-v4-flash:free"  # бесплатная версия
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

print("Загрузка модели эмбеддингов (multilingual-e5-small)...")
embeddings = HuggingFaceEmbeddings(
    model_name="intfloat/multilingual-e5-small",
    model_kwargs={'device': 'cpu'},
    encode_kwargs={'normalize_embeddings': True}
)
print("✅ Модель эмбеддингов готова")
```

```python
from sentence_transformers import CrossEncoder

print("Загрузка кросс-энкодера для реранжинга (ms-marco-MiniLM-L-6-v2)...")
reranker = CrossEncoder('cross-encoder/ms-marco-MiniLM-L-6-v2', device='cpu')
print("✅ Кросс-энкодер загружен")
```

```python
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import Chroma

# Базовые документы о политике командировок
docs = [
    Document(page_content="""Политика компании в отношении авиаперелётов:
    - Эконом-класс для перелётов до 4 часов.
    - Бизнес-класс разрешён при перелётах более 6 часов или для сотрудников уровня директор и выше.
    - Максимальная стоимость билета не должна превышать 50 000 рублей в одну сторону.
    - Рекомендуется выбирать прямые рейсы.""", metadata={"source": "policy"}),

    Document(page_content="""Политика компании в отношении проживания:
    - Суточные: 2500 рублей в день по России, 5000 рублей за границей.
    - Отели: не более 8000 рублей/сутки в Москве, 12000 рублей/сутки за рубежом.
    - Предпочтение отдаётся отелям с завтраком и бесплатным Wi-Fi.""", metadata={"source": "policy"}),

    Document(page_content="""Процесс оформления командировки:
    1. Сотрудник подаёт заявку не менее чем за 5 рабочих дней до выезда.
    2. Руководитель согласует командировку.
    3. После согласования помощник ищет билеты и отель в рамках бюджета.
    4. По возвращении сдаётся авансовый отчёт.""", metadata={"source": "process"})
]

# Чанкинг
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
    separators=["\n\n", "\n", ". ", " ", ""]
)
chunks = splitter.split_documents(docs)
print(f"Создано {len(chunks)} чанков")

# Создание векторной БД (Chroma)
vectorstore = Chroma.from_documents(chunks, embeddings)
print("✅ Векторная база данных создана")
```

```python
def retrieve_relevant_chunks(query: str, k: int = 5, top_n: int = 3) -> List[str]:
    """
    1. Поиск по векторной БД (k кандидатов).
    2. Реранжинг с помощью кросс-энкодера.
    3. Возврат top_n самых релевантных текстов.
    """
    # Шаг 1: Поиск кандидатов
    candidates = vectorstore.similarity_search(query, k=k)
    candidate_texts = [doc.page_content for doc in candidates]

    # Шаг 2: Реранжинг (кросс-энкодер оценивает пары (запрос, текст))
    pairs = [(query, text) for text in candidate_texts]
    scores = reranker.predict(pairs)  # возвращает список float

    # Сортируем по убыванию score
    ranked = sorted(zip(candidate_texts, scores), key=lambda x: x[1], reverse=True)

    # Шаг 3: Берём top_n
    return [text for text, _ in ranked[:top_n]]
```

```python
from langgraph.graph import StateGraph, END

# Состояние, которое передаётся между агентами
class AgentState(TypedDict):
    user_query: str                # исходный запрос
    extracted_params: dict         # извлечённые параметры (город, даты и т.д.)
    policy_context: str            # результат RAG-поиска (правила)
    tickets: List[dict]            # найденные билеты
    budget_check: dict             # результат проверки бюджета
    final_answer: str              # итоговый ответ пользователю
    next_agent: str                # следующий агент для вызова

# Имитация API поиска билетов
def mock_ticket_search(departure: str, arrival: str, date: str) -> List[dict]:
    return [
        {"airline": "Аэрофлот", "price": 14500, "flight": "SU1234", "departure_time": "08:00"},
        {"airline": "S7 Airlines", "price": 12800, "flight": "S7001", "departure_time": "14:30"},
        {"airline": "Победа", "price": 8900, "flight": "DP456", "departure_time": "20:15"}
    ]

# Имитация бюджетного аналитика (сравнение с лимитами из политики)
def check_budget(ticket_price: int, policy_rules: str) -> dict:
    # Упрощённо: ищем в policy_rules упоминание лимита
    if "50 000" in policy_rules:
        limit = 50000
    else:
        limit = float('inf')
    is_affordable = ticket_price <= limit
    return {
        "ticket_price": ticket_price,
        "limit": limit,
        "within_budget": is_affordable,
        "message": f"Билет за {ticket_price} руб. {'входит' if is_affordable else 'превышает'} лимит ({limit} руб.)"
    }
```

```python
def extract_params(state: AgentState) -> AgentState:
    """Оркестратор: извлекает параметры из запроса (упрощённо имитируем)."""
    # В реальной системе здесь вызов LLM или rule-based extraction
    query = state["user_query"].lower()
    params = {
        "departure": "Москва" if "москв" in query else "не указан",
        "arrival": "Санкт-Петербург" if "питер" in query or "петербург" in query else "не указан",
        "date": "2025-06-01"
    }
    state["extracted_params"] = params
    state["next_agent"] = "get_policy"
    return state

def get_policy(state: AgentState) -> AgentState:
    """Агент-эксперт по политикам: использует RAG для получения правил."""
    query = state["user_query"]
    relevant_chunks = retrieve_relevant_chunks(query, k=4, top_n=2)
    state["policy_context"] = "\n".join(relevant_chunks)
    state["next_agent"] = "search_tickets"
    return state

def search_tickets(state: AgentState) -> AgentState:
    """Агент-поисковик билетов."""
    params = state["extracted_params"]
    tickets = mock_ticket_search(params["departure"], params["arrival"], params["date"])
    state["tickets"] = tickets
    state["next_agent"] = "check_budget"
    return state

def check_budget_agent(state: AgentState) -> AgentState:
    """Агент-аналитик бюджета."""
    # Берем первый билет для простоты (в реальности можно проверять все)
    if state["tickets"]:
        ticket = state["tickets"][0]
        budget_result = check_budget(ticket["price"], state["policy_context"])
        state["budget_check"] = budget_result
    else:
        state["budget_check"] = {"within_budget": False, "message": "Билеты не найдены"}
    state["next_agent"] = "generate_response"
    return state

def generate_response(state: AgentState) -> AgentState:
    """Формирует итоговый ответ (можно вызвать LLM через OpenRouter для улучшения)."""
    # Если нужен красивый ответ с LLM – используем OpenRouter.
    # Но для демонстрации можно и сформировать шаблон.
    if state["tickets"]:
        ticket = state["tickets"][0]
        budget = state["budget_check"]
        answer = f"""
По вашему запросу: "{state['user_query']}"

Найден билет: {ticket['airline']} рейс {ticket['flight']}, цена {ticket['price']} руб.
Проверка бюджета: {budget['message']}

Рекомендация по политике компании:
{state['policy_context'][:200]}...
"""
    else:
        answer = "Не удалось найти билеты по вашему запросу."

    state["final_answer"] = answer
    state["next_agent"] = "end"
    return state

def router(state: AgentState) -> str:
    """Определяет, куда перейти дальше на основе поля next_agent."""
    if state["next_agent"] == "end":
        return END
    return state["next_agent"]
```

```python
# Создаём граф
workflow = StateGraph(AgentState)

# Добавляем узлы
workflow.add_node("extract_params", extract_params)
workflow.add_node("get_policy", get_policy)
workflow.add_node("search_tickets", search_tickets)
workflow.add_node("check_budget", check_budget_agent)
workflow.add_node("generate_response", generate_response)

# Устанавливаем точку входа
workflow.set_entry_point("extract_params")

# Добавляем условные рёбра
workflow.add_conditional_edges("extract_params", router)
workflow.add_conditional_edges("get_policy", router)
workflow.add_conditional_edges("search_tickets", router)
workflow.add_conditional_edges("check_budget", router)
workflow.add_conditional_edges("generate_response", router)

# Компилируем
app = workflow.compile()
print("✅ Мультиагентная система готова")
```

```python
# Пример запроса
test_query = "Нужны билеты из Москвы в Питер на завтра, какие правила по бюджету?"

initial_state = {
    "user_query": test_query,
    "extracted_params": {},
    "policy_context": "",
    "tickets": [],
    "budget_check": {},
    "final_answer": "",
    "next_agent": ""
}

result = app.invoke(initial_state)
print("=== ИТОГОВЫЙ ОТВЕТ ===\n")
print(result["final_answer"])
```
